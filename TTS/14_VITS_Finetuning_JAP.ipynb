```python
# 二：测试相关库
# re：正则
# → 文本清理

# jaconv
# → 日语字符形式转换：
# jaconv.hira2kata("こんにちは")，变成
# コンニチハ

# pyopenjtalk：
# 日语 G2P（grapheme即字形字符化成phoneme），把 Text 转成 Phoneme sequence


# unicodedata：日语假名用全角，英文和数字用半角。
# → 日语 Text → Phoneme
# 假名：统一成全角
# ｶﾀｶﾅ → カタカナ

# 英文/数字：统一成半角
# ＡＢＣ → ABC
# １２３ → 123
import re
import unicodedata
import pyopenjtalk
import jaconv
```

```python
# 1. unicodedata：字符规范化
# NFKC 是 Unicode 的一种规范化方式，全称是：
# Normalization Form Compatibility Composition
# 你可以先不记全称，记它的作用就行：
# 把“看起来一样但编码/形式不同”的字符，尽量统一成标准形式。
text1 = "ｶﾀｶﾅ ＡＢＣ ２０２６"
print("unicodedata:")
print(unicodedata.normalize("NFKC", text1))
```

```python
# 2. pyopenjtalk：日语 Text → Phoneme
text2 = "こんにちは、今日はいい天気ですね。"
print("\npyopenjtalk:")
print(pyopenjtalk.g2p(text2))
```

```python
# 3. jaconv：假名 / 全角半角转换
text3 = "こんにちは"
print("\njaconv:")
print(jaconv.hira2kata(text3))
```

```python
# 三：建立normalization函数预处理日语text

# 这一步： text = re.sub(r"\s+", " ", text)，为什么不全去掉空格？
# 原因是有些文本里，空格本身可能有分隔作用，尤其是：
# 日本語 English 123
# 如果你全部删掉，就会变成：
# 日本語English123
# 这样反而可能破坏文本结构。

def normalize_japanese_text(text):

    # 半角假名 → 全角假名（h2z表示：half-width to full-width）
    text = jaconv.h2z(
        text,
        kana=True,#true表示转化假名
        ascii=False,#ascii值ascii字符，一般指英语和一些符号，比如@)，false表示不要把它化成全角
        digit=False #不要把数字化成全角
    )

    # 进一步用Unicode 规范化
    text = unicodedata.normalize("NFKC", text)

    # 多个连续空格 → 一个空格
    text = re.sub(r"\s+", " ", text)

    # 删除句首、句尾空格
    text = text.strip()

    return text
```

```python
# 把音频重采样到22050 Hz
import os
import librosa
import soundfile as sf

output_root = "/content/hadou_22050"

def resample_to_22050(example):
    # 原始 48 kHz wav
    src = example["audio_path"]

    # 保留原来的目录结构
    # 例如 audio/aica/AICA_0001.wav
    rel_path = example["file_name"]

    # 新的 22050 Hz 文件路径
    dst = os.path.join(output_root, rel_path)

    # 创建对应文件夹
    os.makedirs(os.path.dirname(dst), exist_ok=True)

    # 读取 + 重采样到 22050 Hz
    y, sr = librosa.load(
        src,
        sr=22050,
        mono=True
    )

    # 真正保存成新的 22050 Hz wav 文件
    sf.write(dst, y, 22050)

    # 在 dataset 中记录新路径
    example["audio_path_22050"] = dst

    return example


dataset = dataset.map(resample_to_22050)
```

```python
# 因为 Hugging Face Dataset 的结构不能直接被 ESPnet/VITS 读取，所以要先转换成它规定的 text 和 wav.scp 格式。
# 这样模型训练时才能根据每个 utterance ID 正确对应到文本内容和音频文件路径。
# # text
# A001 こんにちは

# # wav.scp
# A001 /path/to/A001.wav
```

```python
#八：把处理好的 Hadou dataset 转化成ESPnet/VITS需要的输入的基本格式
import os

# 最终生成 ESPnet 训练数据文件的总目录
espnet_data_root = "/content/hadou_espnet"


def write_espnet_files(ds, split_name):
    """
    把 Hugging Face Dataset 转换成 ESPnet 需要的两个文件：
    1. text     ：utterance ID + 文本
    2. wav.scp  ：utterance ID + 音频路径
    """

    # 根据 train / valid / test 创建对应文件夹
    # 例如：
    # /content/hadou_espnet/train
    # /content/hadou_espnet/valid
    # /content/hadou_espnet/test
    split_dir = os.path.join(espnet_data_root, split_name)
    os.makedirs(split_dir, exist_ok=True)

    # ESPnet 需要的文本文件
    text_path = os.path.join(split_dir, "text")

    # ESPnet 需要的音频路径文件
    wav_scp_path = os.path.join(split_dir, "wav.scp")

    # 同时打开 text 和 wav.scp，准备写入
    with open(text_path, "w", encoding="utf-8") as f_text, \
         open(wav_scp_path, "w", encoding="utf-8") as f_wav:

        # 遍历当前 train / valid / test 中的每一条数据
        for example in ds:

            # 每条语音的唯一 ID
            # 例如 AICA_0001
            utt_id = example["id"]

            # 这条语音对应的日语文本
            text = example["text"]

            # 已经重采样为 22050 Hz 的 wav 文件路径
            wav_path = example["audio_path_22050"]

            # 写入 text
            # 最终格式：
            # AICA_0001 こんにちは。
            f_text.write(f"{utt_id} {text}\n")

            # 写入 wav.scp
            # 最终格式：
            # AICA_0001 /content/hadou_22050/audio/aica/AICA_0001.wav
            f_wav.write(f"{utt_id} {wav_path}\n")


# 把训练集转换成 ESPnet 格式
write_espnet_files(train_dataset, "train")

# 把验证集转换成 ESPnet 格式
write_espnet_files(val_dataset, "valid")

# 把测试集转换成 ESPnet 格式
write_espnet_files(test_dataset, "test")
```

```python
#九：转化成ESPnet/VITS训练需要的格式：即补充 speaker 映射信息

#发现上面输出结果正好是
# text      ：ID → 文本
# wav.scp   ：ID → 音频
# utt2spk   ：ID → 说话人
# spk2utt   ：说话人 → 所有ID


# ESPnet/VITS 的训练流程要求数据先整理成它规定的输入格式，即
# 生成 utt2spk：为每条 utterance 写上对应的 speaker，例如 AICA_0001 Hadou。
# 原因：ESPnet 需要知道每条语音属于哪个 speaker。

# 生成 spk2utt：把 Hadou 对应到所有 utterance ID。
# 原因：这是 utt2spk 的反向索引，ESPnet/Kaldi 的标准数据目录通常会保留它。

# - utt2spk：从“语音 ID”找到“说话人”
# - spk2utt：从“说话人”反过来找到“他有哪些语音”
# 你这里只有一个人 Hadou，所以本质上就是给所有音频统一标记成同一个 speaker。这样 ESPnet 后面处理数据时，就能按它的标准格式正常读取
```

```python
import os

espnet_data_root = "/content/hadou_espnet"

def prepare_espnet_split(split_name):
    split_dir = os.path.join(espnet_data_root, split_name)

    text_path = os.path.join(split_dir, "text")
    wav_path = os.path.join(split_dir, "wav.scp")
    utt2spk_path = os.path.join(split_dir, "utt2spk")
    spk2utt_path = os.path.join(split_dir, "spk2utt")

    # ---------- 1. 读取并排序 text ----------
    with open(text_path, "r", encoding="utf-8") as f:
        text_lines = f.readlines()

    text_lines = sorted(
        text_lines,
        key=lambda x: x.split(maxsplit=1)[0]
    )

    with open(text_path, "w", encoding="utf-8") as f:
        f.writelines(text_lines)

    # ---------- 2. 读取并排序 wav.scp ----------
    with open(wav_path, "r", encoding="utf-8") as f:
        wav_lines = f.readlines()

    wav_lines = sorted(
        wav_lines,
        key=lambda x: x.split(maxsplit=1)[0]
    )

    with open(wav_path, "w", encoding="utf-8") as f:
        f.writelines(wav_lines)

    # ---------- 3. 创建 utt2spk ----------
    utt_ids = [
        line.split(maxsplit=1)[0]
        for line in text_lines
    ]

    with open(utt2spk_path, "w", encoding="utf-8") as f:
        for utt_id in utt_ids:
            f.write(f"{utt_id} hadou\n")

    # ---------- 4. 创建 spk2utt ----------
    with open(spk2utt_path, "w", encoding="utf-8") as f:
        f.write("hadou " + " ".join(utt_ids) + "\n")

    print(
        split_name,
        ":", len(utt_ids),
        "utterances"
    )


for split in ["train", "valid", "test"]:
    prepare_espnet_split(split)
```

# 十：开始 VITS Fine-tuning（Python 风格）

下面开始正式微调。

这一部分不再使用 `!python -m espnet2.bin.gan_tts_train ...` 的 Shell 写法，
而是直接使用 ESPnet 的 Python API：`GANTTSTask.main()`。

整体流程：

`加载预训练模型 → 建立微调配置 → 检查配置 → 定义数据与训练参数 → 开始训练`


```python
# 1. 加载预训练 VITS checkpoint
# 作用：
# 下载 JSUT 的预训练 VITS，并找到原始 config 和模型权重。
# 后面的 Hadou 微调会从这些预训练参数继续训练。

from huggingface_hub import snapshot_download
from pathlib import Path

# Hugging Face 上的预训练 VITS
MODEL_ID = "espnet/kan-bayashi_jsut_vits_prosody"

# 下载模型；如果之前已经下载过，会直接使用本地缓存
model_root = Path(
    snapshot_download(MODEL_ID)
)

# ESPnet 模型所在目录
pretrained_dir = (
    model_root
    / "exp"
    / "tts_train_vits_raw_phn_jaconv_pyopenjtalk_prosody"
)

# 原始模型配置
pretrained_config = pretrained_dir / "config.yaml"

# 预训练 VITS 权重
pretrained_model = (
    pretrained_dir
    / "train.total_count.ave_10best.pth"
)

# 确认文件存在
print("Config:", pretrained_config)
print("Model:", pretrained_model)
print("Config exists:", pretrained_config.exists())
print("Model exists:", pretrained_model.exists())

assert pretrained_config.exists(), "找不到预训练 config.yaml"
assert pretrained_model.exists(), "找不到预训练模型权重"

# 一句话总结：
# 这段代码取得 JSUT 预训练 VITS 的配置和权重，作为 Hadou 微调的起点。
```

## 为什么继续使用原来的 config？

VITS 的 checkpoint 和训练时的网络结构必须对应。

所以这里不从空白重新写配置，而是：

`原 JSUT config → 保留 VITS 网络结构 → 只修改 fine-tuning 参数`

这样可以尽量保证预训练权重和当前模型结构一致。


```python
# 2. 创建 Hadou 的 fine-tuning 配置
# 作用：
# 读取预训练 VITS 原来的 config，只修改学习率、batch size、训练长度等参数。
# 同时删除当前 ESPnet 不认识的旧版配置字段，例如 required / version。
#
# 重要：训练长度只在这一格定义一次。
# 以后如果要从 2 epoch 改成 10 epoch，只改这里并从这一格往下重新运行。

import yaml
from espnet2.tasks.gan_tts import GANTTSTask

# =========================
# 本次实验统一训练参数
# =========================
MAX_EPOCH = 20
ITERS_PER_EPOCH = 100
LOG_INTERVAL = 5

# False：从 JSUT 预训练权重重新开始 fine-tuning
# True ：从 /content/hadou_vits_finetune/checkpoint.pth 接着训练
RESUME_TRAINING = False

# 读取预训练模型原来的配置
with open(pretrained_config, "r", encoding="utf-8") as f:
    config = yaml.safe_load(f)


# --------------------------------------------------
# 删除不能直接沿用的旧训练信息
# --------------------------------------------------

remove_keys = [
    # 原 JSUT 数据路径
    "train_data_path_and_name_and_type",
    "valid_data_path_and_name_and_type",
    "train_shape_file",
    "valid_shape_file",

    # 原输出目录 / 初始化参数
    "output_dir",
    "init_param",

    # 旧版 ESPnet 的元信息
    "required",
    "version",

    # 原来的分布式训练信息
    "dist_world_size",
    "dist_rank",
    "local_rank",
    "dist_master_addr",
    "dist_master_port",
    "dist_launcher",
    "distributed",
]

for key in remove_keys:
    config.pop(key, None)


# --------------------------------------------------
# 设置 Hadou fine-tuning 参数
# --------------------------------------------------

# 与后面的 train_args 使用同一组变量，避免多次运行后数值不一致
config["max_epoch"] = MAX_EPOCH
config["num_iters_per_epoch"] = ITERS_PER_EPOCH

# VITS 有 Generator 和 Discriminator 两个 optimizer
config["optim_conf"]["lr"] = 1e-4
config["optim2_conf"]["lr"] = 1e-4

# 小数据集先使用较小 batch
config["batch_type"] = "unsorted"
config["valid_batch_type"] = "unsorted"
config["batch_size"] = 4
config["valid_batch_size"] = 4


# 这两个都不是模型结构参数，而是训练运行/保存相关的配置。
# - config["num_workers"] = 2：DataLoader 用 2 个子进程并行读取和预处理训练数据。主要影响数据加载速度，不直接改变 VITS 的网络结构和 loss。设太大可能占更多 CPU/RAM；Colab 里 2 通常是比较稳妥的值。
# - config["keep_nbest_models"] = 5：训练过程中只保留表现最好的 5 个 checkpoint。这里的 “best” 通常根据 ESPnet 配置中的验证集指标/排序规则决定，不是简单说“最后 5 个 epoch”。


# DataLoader worker 数
config["num_workers"] = 2

# 保存表现较好的 checkpoint
config["keep_nbest_models"] = 5

# Colab 使用 1 张 GPU
config["ngpu"] = 1


# --------------------------------------------------
# 自动检查当前 ESPnet 能否识别 config 中的字段
# --------------------------------------------------

parser = GANTTSTask.get_parser()

valid_keys = {
    action.dest
    for action in parser._actions
}

unknown_keys = sorted(
    key
    for key in config.keys()
    if key not in valid_keys
)

if unknown_keys:
    print("删除当前 ESPnet 不识别的旧配置字段：")
    print(unknown_keys)

    for key in unknown_keys:
        config.pop(key, None)
else:
    print("没有发现额外的不兼容配置字段。")


# --------------------------------------------------
# 保存新的 fine-tuning config
# --------------------------------------------------

FINETUNE_CONFIG = Path(
    "/content/hadou_vits_finetune.yaml"
)

with open(FINETUNE_CONFIG, "w", encoding="utf-8") as f:
    yaml.safe_dump(
        config,
        f,
        allow_unicode=True,
        sort_keys=False
    )

print("\nFine-tuning config:")
print(FINETUNE_CONFIG)
print(f"MAX_EPOCH = {MAX_EPOCH}")
print(f"ITERS_PER_EPOCH = {ITERS_PER_EPOCH}")
print(f"LOG_INTERVAL = {LOG_INTERVAL}")
print(f"RESUME_TRAINING = {RESUME_TRAINING}")

# 一句话总结：
# 这一格是训练长度的唯一来源，避免 YAML、变量和 train_args 互相不一致。
```

```python
# 6. 可选：先做 10-step smoke test
# 作用：
# 正式训练之前只跑 10 个 iteration，确认数据读取、G2P、batching、
# pretrained weights、Generator / Discriminator 和 loss 都能正常工作。
#
# 第一次干净重跑建议保持 True。
# smoke test 永远不 resume，并使用单独目录，不会污染正式训练 checkpoint。

RUN_SMOKE_TEST = True

if RUN_SMOKE_TEST:
    smoke_test_args = train_args.copy()

    # 不追加重复参数，而是直接替换 train_args 中已有的值
    smoke_overrides = {
        "--output_dir": SMOKE_OUTPUT_DIR,
        "--max_epoch": "1",#只跑一个epoch
        "--num_iters_per_epoch": "10",
        "--resume": "false",#每次都重头，从与训练模型开始，不从之前的checkpoint开始
    }

    for name, value in smoke_overrides.items():
        idx = smoke_test_args.index(name)
        smoke_test_args[idx + 1] = value

    print("开始 10-step smoke test...")
    GANTTSTask.main(
        cmd=smoke_test_args
    )
    print("10-step smoke test 完成。")
else:
    print("跳过 smoke test。")

# 一句话总结：
# 这一格只验证训练 pipeline；正式训练仍使用 OUTPUT_DIR。
```

```python
# 7. 正式开始 VITS fine-tuning
# 作用：
# 1) 强制显示 ESPnet INFO 日志（其中包含 loss）
# 2) 根据日志更新 tqdm 进度条
# 3) 如果以后改为 resume，会自动按“剩余 epoch”计算进度，而不是错误地显示总历史步数

import logging
import re
import torch
from pathlib import Path
from tqdm.auto import tqdm


# =========================
# 确保 Colab 显示 INFO 日志
# =========================

root_logger = logging.getLogger()
root_logger.setLevel(logging.INFO)

for handler in root_logger.handlers:
    handler.setLevel(logging.INFO)


# =========================
# 计算本次真正需要训练多少 iteration
# =========================

START_EPOCH = 0
checkpoint_path = Path(OUTPUT_DIR) / "checkpoint.pth"

if RESUME_TRAINING and checkpoint_path.exists():
    old_checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=False
    )
    START_EPOCH = int(
        old_checkpoint.get("reporter", {}).get("epoch", 0)
    )
    del old_checkpoint

REMAINING_EPOCHS = max(MAX_EPOCH - START_EPOCH, 0)
TOTAL_ITERS = REMAINING_EPOCHS * ITERS_PER_EPOCH


# =========================
# 创建进度条
# =========================

pbar = tqdm(
    total=TOTAL_ITERS,
    desc="VITS fine-tuning",
    unit="iter"
)


# =========================
# 从 ESPnet INFO 日志读取训练进度和 loss
# =========================

class ESPnetProgressHandler(logging.Handler):

    def emit(self, record):
        msg = record.getMessage()

        # ESPnet Reporter 日志形式：
        # 1epoch:train:1-5batch: loss=..., ...
        match = re.search(
            r"(\d+)epoch:train:(\d+)(?:-(\d+))?batch",
            msg
        )

        if not match:
            return

        epoch = int(match.group(1))
        start_iter = int(match.group(2))
        end_iter = (
            int(match.group(3))
            if match.group(3)
            else start_iter
        )

        # 相对“本次运行”的进度；resume 时不会把旧 epoch 重复算进去
        completed = (
            (epoch - START_EPOCH - 1) * ITERS_PER_EPOCH
            + min(end_iter, ITERS_PER_EPOCH)
        )

        pbar.n = max(0, min(completed, TOTAL_ITERS))

        # 自动抓取日志里名字包含 loss 的指标，最多显示前 4 个
        # 正常 INFO 日志本身也会继续显示在 Cell 下方
        loss_pairs = re.findall(
            r"([A-Za-z0-9_]*loss)=([+-]?(?:\d+(?:\.\d*)?|\.\d+)(?:[eE][+-]?\d+)?)",
            msg
        )

        if loss_pairs:
            postfix = {
                key: value
                for key, value in loss_pairs[:4]
            }
            pbar.set_postfix(postfix, refresh=False)

        pbar.refresh()


progress_handler = ESPnetProgressHandler()
progress_handler.setLevel(logging.INFO)
root_logger.addHandler(progress_handler)


# =========================
# 开始训练
# =========================

print("=" * 50)
print("开始 VITS fine-tuning")
print(f"Checkpoint 已训练到: {START_EPOCH} epoch")
print(f"目标 Max epoch: {MAX_EPOCH}")
print(f"本次需要训练: {REMAINING_EPOCHS} epoch")
print(f"Iterations per epoch: {ITERS_PER_EPOCH}")
print(f"本次预计最多训练: {TOTAL_ITERS} iterations")
print("=" * 50)

try:
    if TOTAL_ITERS == 0:
        print(
            f"当前 checkpoint 已达到 epoch {START_EPOCH}，"
            f"目标 max_epoch={MAX_EPOCH}，因此没有新的训练需要执行。"
        )
    else:
        GANTTSTask.main(
            cmd=train_args
        )

finally:
    root_logger.removeHandler(progress_handler)
    pbar.close()

print("=" * 50)
print("VITS fine-tuning Cell 执行结束")
print("=" * 50)
```

```python
# 8. 检查最终 checkpoint
# 作用：确认真正训练到哪个 epoch，并查看最后一个 epoch 的 loss 摘要。

import os
import torch

checkpoint_path = f"{OUTPUT_DIR}/checkpoint.pth"

if not os.path.exists(checkpoint_path):
    print("还没有生成正式训练 checkpoint：", checkpoint_path)
else:
    checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=False
    )

    reporter_state = checkpoint.get("reporter", {})
    final_epoch = reporter_state.get("epoch")

    print("已训练到 epoch:", final_epoch)

    # Reporter 会保存每个 epoch 的 train / valid 统计
    stats = reporter_state.get("stats", {})
    epoch_stats = stats.get(final_epoch, stats.get(str(final_epoch), {}))

    if isinstance(epoch_stats, dict):
        for phase in ("train", "valid"):
            phase_stats = epoch_stats.get(phase, {})
            if isinstance(phase_stats, dict):
                loss_stats = {
                    k: v
                    for k, v in phase_stats.items()
                    if "loss" in str(k).lower()
                }
                if loss_stats:
                    print(f"\n{phase} loss summary:")
                    for k, v in loss_stats.items():
                        print(f"  {k}: {v}")
```

```python
# VITS Loss
# │
# ├── ① Generator Loss = 43.7753
# │      ├── Mel Loss          25.1804
# │      ├── KL Loss            3.0103
# │      ├── Duration Loss      2.8733
# │      ├── Adversarial Loss   2.8631
# │      └── Feature Match Loss 9.8482
# │
# └── ② Discriminator Loss = 2.3753
#        ├── Real Loss          1.2747
#        └── Fake Loss          1.1006


# Mel       → 声音本身对不对
# Duration  → 长度对不对
# KL        → latent分布对不对
# Adv       → 听起来真不真
# FeatMatch → 细节像不像真语音


# 配置里我们看到过类似：
# lambda_mel = 45lambda_kl = 1lambda_dur = 1lambda_adv = 1lambda_feat_match = 2


# 这里的 lambda 是各项 loss 的权重。
# 概念上公式是：
# \[
# L_G =
# \lambda_{mel}L_{mel}
# +\lambda_{kl}L_{kl}
# +\lambda_{dur}L_{dur}
# +\lambda_{adv}L_{adv}
# +\lambda_{fm}L_{fm}
# \]





# generator_adv_loss = Generator Adversarial Loss（生成器对抗损失）
# 它来自 GAN 的对抗训练：生成器希望生成的语音能让 Discriminator 判断成“真实语音”。
# 你可以简单记成：
# adv loss = 生成语音“像不像真的”这一项 loss。
# 不要只盯着 total generator loss 越低越好。
# 因为 Generator 和 Discriminator 是互相博弈的，它们的 adversarial loss 本来就可能波动。



# 第一优先：
# generator_mel_loss
# → 生成声音是否越来越接近真实声音

# 第二优先：
# generator_dur_loss
# → 发音长度 / 节奏有没有学好

# 第三优先：
# generator_adv_loss + discriminator_loss
# → GAN训练有没有失衡
```

```python
#  GAN 的两个组件
# - Generator（生成器）：负责“造假”。在你的 VITS 里，它根据文本/latent 等信息生成语音波形。
# - Discriminator（判别器）：负责“鉴假”。它看真实语音和生成语音，判断哪个是真的、哪个是生成的。
# 在 VITS 里可以简单理解成：
# 文本
#  ↓
# Generator
#  ↓
# 生成语音
#  ↓
# Discriminator
#  ↓
# 判断：真 / 假
```

```python
# ============================================================
# Step 2：测试 ESPnet 自动生成的 5-best averaged model
#
# 作用：
# ESPnet 会把多个 checkpoint 的参数进行平均，
# 得到一个 averaged model。
#
# 参数平均有时会让模型：
# - 更稳定
# - 减少单个 epoch 的波动
# - 推理效果稍微更好
#
# 这里使用和 Step 1 完全相同的文本，
# 方便直接比较：
#
# 20epoch.pth
# VS
# train.total_count.ave_5best.pth
# ============================================================

import torch
from espnet2.bin.tts_inference import Text2Speech
from IPython.display import Audio, display


# -------------------------
# 模型文件路径
# -------------------------

CONFIG_PATH = "/content/hadou_vits_finetune/config.yaml"

MODEL_PATH_AVG = (
    "/content/hadou_vits_finetune/"
    "train.total_count.ave_5best.pth"
)


# -------------------------
# 自动选择 GPU / CPU
# -------------------------

device = "cuda" if torch.cuda.is_available() else "cpu"


# -------------------------
# 加载 averaged VITS 模型
# -------------------------

text2speech_avg = Text2Speech(
    train_config=CONFIG_PATH,
    model_file=MODEL_PATH_AVG,
    device=device,
)


# -------------------------
# 和 Step 1 使用完全相同的文本
# -------------------------

text = "こんにちは。今日はとてもいい天気ですね。"


# -------------------------
# 执行 TTS 推理
# -------------------------

with torch.no_grad():
    output_avg = text2speech_avg(text)


# -------------------------
# 获取 waveform
# -------------------------

wav_avg = output_avg["wav"].cpu().numpy()


# -------------------------
# 播放 averaged model 的结果
# -------------------------

print("Model: train.total_count.ave_5best.pth")
print("Device:", device)
print("Sampling rate:", text2speech_avg.fs)

display(
    Audio(
        wav_avg,
        rate=text2speech_avg.fs
    )
)
```

```python
#我觉得checkpoints20的模型更好一点，上传
# ============================================================
# Step 3：整理 Hugging Face 发布文件
#
# 这里选择第 20 个 epoch 的模型：
# /content/hadou_vits_finetune/20epoch.pth
#
# 然后复制并重命名为：
# /content/hadou_vits_release/model.pth
# ============================================================

import shutil
from pathlib import Path


# 创建发布文件夹
RELEASE_DIR = Path("/content/hadou_vits_release")
RELEASE_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# 选择 epoch 20 的模型参数
SOURCE_MODEL = "/content/hadou_vits_finetune/20epoch.pth"


# 把 20epoch.pth 复制成更通用的名字 model.pth
shutil.copy(
    SOURCE_MODEL,
    RELEASE_DIR / "model.pth"
)


# 同时复制配置文件
shutil.copy(
    "/content/hadou_vits_finetune/config.yaml",
    RELEASE_DIR / "config.yaml"
)


# 检查最终准备上传的文件
print("Release files:")

for p in RELEASE_DIR.iterdir():
    print(p)
```
