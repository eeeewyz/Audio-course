```python
from datasets import load_dataset
#用streaming传输数据集
voxpopuli_croatian = load_dataset("facebook/voxpopuli", "en",streaming=True)
```

```python
#保存iterator
iterator=iter(voxpopuli_croatian["train"])
```

```python
#打印三条样本
sample1=next(iterator)
sample2=next(iterator)
sample3=next(iterator)
print(sample1)
print(sample2)
print(sample3)
```

```python
#绘制waveform
!pip install librosa
audio = sample1["audio"]

array = audio.get_all_samples().data.numpy().squeeze()
sampling_rate = audio.get_all_samples().sample_rate
```

```python
import matplotlib.pyplot as plt
import librosa.display

plt.figure(figsize=(12, 4))
librosa.display.waveshow(array, sr=sampling_rate)
plt.show()
```
